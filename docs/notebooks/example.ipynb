# Scoring a preprocessing run

Which metrics can actually rank a method, and which only describe one?

```python
import sprotmetrics as spm

catalogue = spm.list_metrics()
catalogue[["label", "task", "direction", "strength"]].head(10)
```

`maximize` is `None` wherever ranking on a metric would be unsound, so the
rankable subset is explicit rather than assumed:

```python
rankable = spm.list_metrics(directional_only=True)
print(f"{len(rankable)} of {len(catalogue)} metrics have a usable direction")
rankable.groupby("task")["maximize"].count()
```

Metrics the package records but deliberately does not compute keep their
reasoning attached, so a redundancy stays documented rather than rediscovered:

```python
retired = spm.list_metrics(include_deprecated=True).query("superseded_by != ''")
retired[["superseded_by", "notes"]]
```
